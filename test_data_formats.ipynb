```python
from pathlib import Path
from data_models import Record
```

```python
# ============================================================
#  CONFIGURAÇÃO DOS CAMINHOS
# ============================================================
BASE_PATH = Path("data")

XML_PATH = BASE_PATH / "SemClinBr" / "annotated_records"
ARGILLA_PATH = BASE_PATH / "Spesia" / "argilla" / "annotated_records"
DOCCANO_PATH = BASE_PATH / "Spesia" / "doccano" / "annotated_records"
```

```python
# ============================================================
# FUNÇÃO AUXILIAR DE TESTE
# ============================================================
def test_format(name: str, file_path: Path):
    print(f"\n{'='*80}")
    print(f"Testando formato: {name}")
    print(f"Arquivo: {file_path.name}")
    print(f"{'-'*80}")

    try:
        records = Record.from_file(file_path)
        print(f" Total de registros carregados: {len(records)}")

        if len(records) > 0:
            r = records[0]
            print(f"Texto (primeiros 120 chars): {r.text[:120]!r}")
            print(f"Total de anotações: {len(r.annotations)}")

            # Mostra detalhes da primeira anotação
            if r.annotations:
                a = r.annotations[0]
                print(f"Anotação exemplo → texto='{a.text}', tags={a.tags}, start={a.start}, end={a.end}")

            # Mostra as tags distintas do documento
            print(f"Tags distintas neste record: {sorted(list(r.tags))[:15]}")
        else:
            print(" Nenhum record carregado.")

    except Exception as e:
        print(f" Erro ao carregar {file_path.name}: {e}")
```

```python
# ============================================================
# TESTE 1 – XML (SemClinBr)
# ============================================================
xml_files = list(XML_PATH.glob("*.xml"))
if xml_files:
    test_format("XML (SemClinBr)", xml_files[0])
else:
    print("Nenhum arquivo XML encontrado em", XML_PATH)
```

```python
# ============================================================
# TESTE 2 – JSON (Argilla)
# ============================================================
json_files = list(ARGILLA_PATH.glob("*.json"))
if json_files:
    test_format("JSON (Argilla)", json_files[0])
else:
    print("Nenhum arquivo JSON encontrado em", ARGILLA_PATH)
```

```python
# ============================================================
# TESTE 3 – JSONL (Doccano)
# ============================================================
jsonl_files = list(DOCCANO_PATH.glob("*.jsonl"))
if jsonl_files:
    test_format("JSONL (Doccano)", jsonl_files[0])
else:
    print("Nenhum arquivo JSONL encontrado em", DOCCANO_PATH)
```
